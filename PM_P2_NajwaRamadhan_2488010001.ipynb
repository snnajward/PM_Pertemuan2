**C. Langkah Latihan**

**Langkah 1 - Mengenali Fitur dan Label**

```python
import pandas as pd
data = {'luas':[36,70,45,90,54],
        'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}  # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]   # fitur
y = df['harga']                     # label

print(df)
```

**Langkah 2 - Klasifikasi vs regresi (amati tipe label)**

```python
# Label numerik kontinu -> REGRESI
print(df['harga'].tolist())
# Label kategori -> KLASIFIKASI
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']
```

**Langkah 3 — Pembagian data latih & uji**

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))
```
